```python
import pandas as pd

# Δημιουργία λίστας για τη συλλογή των δομημένων δεδομένων
data_list = []

for num, emvada in omades.items():
    athroisma = sum(emvada)
    # Υπολογισμός ωρών για κάθε μία από τις 3 υδροληψίες της ομάδας
    ores = [round(ores_omadas * A / athroisma, 1) for A in emvada]

    # Προσθήκη γραμμής στον πίνακα με τα στοιχεία της ομάδας
    data_list.append({
        "Ομάδα": f"Ομάδα {num}",
        "Υδροληψία 1 (στρ)": emvada[0],
        "Ώρες Υδρ. 1 (h)": ores[0],
        "Υδροληψία 2 (στρ)": emvada[1],
        "Ώρες Υδρ. 2 (h)": ores[1],
        "Υδροληψία 3 (στρ)": emvada[2],
        "Ώρες Υδρ. 3 (h)": ores[2],
        "Συνολικές Ώρες (h)": round(sum(ores), 1)
    })

# Μετατροπή σε DataFrame
df_results = pd.DataFrame(data_list)

# Ρύθμιση εμφάνισης για να φαίνονται όλοι οι αριθμοί όμορφα
pd.set_option('display.precision', 2)

# Εμφάνιση του πίνακα
display(df_results)
```

```python
import math

def d_arx(Q, U):
    # D = sqrt(4*Q / (pi*U))
    return math.sqrt((4 * Q) / (math.pi * U))

U_arx = 1.5                                    # αρχική ταχύτητα (m/s)
paroxes = [0.012, 0.024, 0.048, 0.072, 0.12]   # οι 5 διαφορετικές Qmax (m³/s)

# Εισαγωγή των διαμέτρων σε χιλιοστά (mm) από τον κατάλογό σας
katalogος_mm = [
    (110, 99.4),  (125, 113.0), (140, 126.6), (160, 144.6),
    (200, 180.8), (225, 203.4), (250, 226.2), (280, 253.2),
    (315, 285.0), (355, 321.2), (400, 361.8), (450, 407.0),
    (500, 452.2),
]

# Αυτόματη μετατροπή των τιμών από χιλιοστά (mm) σε μέτρα (m)
emporiko_pinakas = [(ex / 1000.0, es / 1000.0) for ex, es in katalogος_mm]

for Q in paroxes:
    D_theoretical = d_arx(Q, U_arx)

    # Επιλογή του καταλληλότερου εμπορικού σωλήνα (πρώτος με εσωτερική διάμετρο >= θεωρητική)
    exot, esot = emporiko_pinakas[-1]  # default τιμή αν καμία δεν κάνει
    for d_ex, d_es in emporiko_pinakas:
        if d_es >= D_theoretical:
            exot, esot = d_ex, d_es
            break

    # Υπολογισμός πραγματικής ταχύτητας: U = Q / (pi * d^2 / 4)
    U = (4 * Q) / (math.pi * (esot ** 2))

    print(f"Q={Q:.3f} m³/s: D_αρχ={D_theoretical*1000:.1f}mm | Επιλογή -> εξωτ={exot*1000:.1f}mm, εσωτ={esot*1000:.1f}mm | Πραγμ. U={U:.2f} m/s")
```

```python
import pandas as pd
import math

# Δημιουργία λίστας για τη συλλογή των αποτελεσμάτων των σωλήνων
pipes_data = []

for Q in paroxes:
    D_theoretical = d_arx(Q, U_arx)

    # Επιλογή καταλληλότερου εμπορικού σωλήνα
    exot, esot = emporiko_pinakas[-1]
    for d_ex, d_es in emporiko_pinakas:
        if d_es >= D_theoretical:
            exot, esot = d_ex, d_es
            break

    # Υπολογισμός πραγματικής ταχύτητας
    U = (4 * Q) / (math.pi * (esot ** 2))

    pipes_data.append({
        "Παροχή Q (m³/s)": Q,
        "Θεωρητική Διάμετρος D (mm)": round(D_theoretical * 1000, 1),
        "Εμπορική Εξωτερική Φ (mm)": round(exot * 1000, 1),
        "Εμπορική Εσωτερική d (mm)": round(esot * 1000, 1),
        "Πραγματική Ταχύτητα U (m/s)": round(U, 2)
    })

# Μετατροπή σε DataFrame και εμφάνιση
df_pipes = pd.DataFrame(pipes_data)
display(df_pipes)
```
