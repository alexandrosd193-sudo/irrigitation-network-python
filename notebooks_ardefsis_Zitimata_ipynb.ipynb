Κανόνας επιλογής: πρώτη εμπορική διάμετρος με **εσωτερική ≥ D_αρχ**, ώστε U ≤ 1.5 m/s.

## Βήμα 1: αρχική διάμετρος από την παροχή
Αναμενόμενο για Q = 0.12: περίπου **319.2 mm**.

```python
import math

def d_arx(Q, U=1.5):
    """Αρχική διάμετρος σε mm από παροχή Q (m³/s) και ταχύτητα U (m/s)."""
    D = math.sqrt(4 * Q / (math.pi * U))   # σε μέτρα
    return D * 1000                         # σε mm

print(d_arx(0.12))
```

## Βήμα 2: κατάλογος εμπορικών διαμέτρων
Ζεύγη (εξωτερική, εσωτερική) σε mm, PVC 10 atm, ταξινομημένα από τη μικρότερη προς τη μεγαλύτερη.

```python
katalogos = [
    (110, 99.4), (125, 110.2), (160, 144.6), (200, 180.8),
    (225, 198.2), (250, 226.8), (280, 246.8), (315, 277.6),
    (355, 321.2), (400, 352.6), (450, 396.6), (500, 440.6),
    (630, 555.2),
]
print(len(katalogos), "εμπορικές διάμετροι")
```

## Βήμα 3: επιλογή διαμέτρου
Περνάει τον κατάλογο με τη σειρά και σταματά στον πρώτο σωλήνα με εσωτερική διάμετρο ≥ D.

```python
def epilogi(D):
    """Επιστρέφει (εξωτ., εσωτ.) της πρώτης διαμέτρου με εσωτ. >= D (mm)."""
    for exot, esot in katalogos:
        if esot >= D:
            return exot, esot
    return None   # αν δεν βρεθεί καμία

print(epilogi(319.2))
```

## Βήμα 4: όλες οι παροχές
Αναμενόμενο:

| Q | εξωτ. | εσωτ. | U |
|---|---|---|---|
| 0.012 | 125 | 110.2 | 1.26 |
| 0.024 | 160 | 144.6 | 1.46 |
| 0.048 | 250 | 226.8 | 1.19 |
| 0.072 | 315 | 277.6 | 1.19 |
| 0.12 | 355 | 321.2 | 1.48 |

```python
paroxes = [0.012, 0.024, 0.048, 0.072, 0.12]

for Q in paroxes:
    D = d_arx(Q)
    exot, esot = epilogi(D)
    U = 4 * Q / (math.pi * (esot / 1000)**2)   # mm -> m
    print(f"Q={Q}: D_αρχ={D:.1f}  εξωτ={exot}  εσωτ={esot}  U={U:.2f} m/s")
```
